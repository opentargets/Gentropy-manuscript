Gene set files for Reactome and KEGG were downloaded from https://maayanlab.cloud/Enrichr/#libraries and merged into one gene_sets.gmt file.

```python
from pathlib import Path

base_dir = Path("..//..//data/gene_sets")
txt_files = sorted(base_dir.glob("*.txt"))
if not txt_files:
    raise FileNotFoundError(f"No gene-set files found in {base_dir}")

merged_lines = []
for txt_file in txt_files:
    with txt_file.open() as fh:
        for line_number, raw_line in enumerate(fh, start=1):
            line = raw_line.rstrip("\n")
            if not line.strip():
                continue
            parts = line.split("\t")
            if len(parts) < 3:
                raise ValueError(f"{txt_file.name}:{line_number} is missing the expected columns (found {len(parts)})")
            name = parts[0].strip()
            if not name:
                raise ValueError(f"{txt_file.name}:{line_number} has an empty gene-set name")
            description = parts[1].strip()
            genes = [gene.strip() for gene in parts[2:] if gene.strip()]
            if not genes:
                raise ValueError(f"{txt_file.name}:{line_number} has no gene symbols after parsing")
            name_with_source = f"{name} {{{txt_file.stem}}}"
            merged_lines.append("\t".join([name_with_source, description, *genes]))

output_path = base_dir / "gene_sets.gmt"
output_path.write_text("\n".join(merged_lines) + "\n")
print(f"Wrote {len(merged_lines)} gene sets to {output_path}")
```
