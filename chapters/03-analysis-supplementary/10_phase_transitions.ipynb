# Supplementary Results 10 — Phase transition probabilities stratified by pleiotropy

Transition rates through Phase I→II, II→III and III→approval for the 18,480 target-indication pairs
carrying genetic support, grouped by the number of therapeutic areas and, separately, by gPS.

Numbers are written to `results/sr10_phase_transitions.json`, and the two tables to
`phase_transition_rates.csv` and `phase_transition_tests.csv` in the derived data directory, which
is what Extended Data Figure 8 plots.

**Provenance.** `manuscript_methods.transitions`, factored out of
`chapters/05-figures-supplementary/extended_data/ed08_translation_success_by_pleiotropy.ipynb`,
whose gPS panel was added in review round 1. Wilson intervals, pairwise proportions z-tests and a
Benjamini-Hochberg adjustment across all nine tests, exactly as the published figure used.

```python
import pandas as pd

from manuscript_methods import paper, transitions

numbers = {}
pairs = pd.read_csv(paper.derived("df_for_enrichment_regression.csv"))
rates, tests = transitions.both_scales(pairs)

numbers["S10.01"] = len(transitions.universe(pairs))
print(f"pairs with genetic support: {numbers['S10.01']:,}")
```

## The two tables

```python
for name in transitions.GROUPINGS:
    print(f"=== {name}")
    wide = rates[name].pivot(index="transition", columns="group", values="rate")
    print(
        (wide[transitions.GROUPINGS[name]["labels"]].reindex([t[2] for t in transitions.TRANSITIONS]) * 100)
        .round(1)
        .to_string()
    )
```

```python
rates_out = pd.concat(rates.values(), ignore_index=True)
tests_out = pd.concat(tests.values(), ignore_index=True)
rates_out.to_csv(paper.derived("phase_transition_rates.csv"), index=False)
tests_out.to_csv(paper.derived("phase_transition_tests.csv"), index=False)
print(f"wrote {len(rates_out)} rate rows and {len(tests_out)} test rows")
tests_out[
    ["grouping", "transition", "comparison", "risk_ratio", "rr_ci_low", "rr_ci_high", "omnibus_p", "p_adj_bh"]
].round(4)
```

```python
IDS = [
    ("S10.01", "universe", None, None, None, "target-indication pairs with genetic support", 18480, 0),
    ("S10.02", "group_n", "TAs", None, 0, "pairs in group 1 TA (therapeutic areas)", 6578, 0),
    ("S10.03", "group_n", "TAs", None, 1, "pairs in group 2–5 TAs (therapeutic areas)", 9705, 0),
    ("S10.04", "group_n", "TAs", None, 2, "pairs in group ≥6 TAs (therapeutic areas)", 2197, 0),
    ("S10.05", "n_at_start", "TAs", "Phase I→II", 0, "Phase I→II, 1 TA: pairs entering", 6578, 0),
    ("S10.06", "rate", "TAs", "Phase I→II", 0, "Phase I→II, 1 TA: success (%)", 82.9, 0.05),
    ("S10.07", "n_at_start", "TAs", "Phase I→II", 1, "Phase I→II, 2–5 TAs: pairs entering", 9705, 0),
    ("S10.08", "rate", "TAs", "Phase I→II", 1, "Phase I→II, 2–5 TAs: success (%)", 84.2, 0.05),
    ("S10.09", "n_at_start", "TAs", "Phase I→II", 2, "Phase I→II, ≥6 TAs: pairs entering", 2197, 0),
    ("S10.10", "rate", "TAs", "Phase I→II", 2, "Phase I→II, ≥6 TAs: success (%)", 82.6, 0.05),
    (
        "S10.11",
        "risk_ratio",
        "TAs",
        "Phase I→II",
        0,
        "Phase I→II, med_vs_low (therapeutic areas): risk ratio",
        1.02,
        0.005,
    ),
    (
        "S10.12",
        "risk_ratio",
        "TAs",
        "Phase I→II",
        1,
        "Phase I→II, high_vs_low (therapeutic areas): risk ratio",
        1.0,
        0.005,
    ),
    (
        "S10.13",
        "risk_ratio",
        "TAs",
        "Phase I→II",
        2,
        "Phase I→II, high_vs_med (therapeutic areas): risk ratio",
        0.98,
        0.005,
    ),
    ("S10.14", "n_at_start", "TAs", "Phase II→III", 0, "Phase II→III, 1 TA: pairs entering", 5453, 0),
    ("S10.15", "rate", "TAs", "Phase II→III", 0, "Phase II→III, 1 TA: success (%)", 57.0, 0.05),
    ("S10.16", "n_at_start", "TAs", "Phase II→III", 1, "Phase II→III, 2–5 TAs: pairs entering", 8173, 0),
    ("S10.17", "rate", "TAs", "Phase II→III", 1, "Phase II→III, 2–5 TAs: success (%)", 53.5, 0.05),
    ("S10.18", "n_at_start", "TAs", "Phase II→III", 2, "Phase II→III, ≥6 TAs: pairs entering", 1815, 0),
    ("S10.19", "rate", "TAs", "Phase II→III", 2, "Phase II→III, ≥6 TAs: success (%)", 49.8, 0.05),
    (
        "S10.20",
        "risk_ratio",
        "TAs",
        "Phase II→III",
        0,
        "Phase II→III, med_vs_low (therapeutic areas): risk ratio",
        0.94,
        0.005,
    ),
    (
        "S10.21",
        "risk_ratio",
        "TAs",
        "Phase II→III",
        1,
        "Phase II→III, high_vs_low (therapeutic areas): risk ratio",
        0.87,
        0.005,
    ),
    (
        "S10.22",
        "risk_ratio",
        "TAs",
        "Phase II→III",
        2,
        "Phase II→III, high_vs_med (therapeutic areas): risk ratio",
        0.93,
        0.005,
    ),
    ("S10.23", "n_at_start", "TAs", "Phase III→approval", 0, "Phase III→approval, 1 TA: pairs entering", 3110, 0),
    ("S10.24", "rate", "TAs", "Phase III→approval", 0, "Phase III→approval, 1 TA: success (%)", 29.4, 0.05),
    ("S10.25", "n_at_start", "TAs", "Phase III→approval", 1, "Phase III→approval, 2–5 TAs: pairs entering", 4376, 0),
    ("S10.26", "rate", "TAs", "Phase III→approval", 1, "Phase III→approval, 2–5 TAs: success (%)", 31.2, 0.05),
    ("S10.27", "n_at_start", "TAs", "Phase III→approval", 2, "Phase III→approval, ≥6 TAs: pairs entering", 903, 0),
    ("S10.28", "rate", "TAs", "Phase III→approval", 2, "Phase III→approval, ≥6 TAs: success (%)", 30.6, 0.05),
    (
        "S10.29",
        "risk_ratio",
        "TAs",
        "Phase III→approval",
        0,
        "Phase III→approval, med_vs_low (therapeutic areas): risk ratio",
        1.06,
        0.005,
    ),
    (
        "S10.30",
        "risk_ratio",
        "TAs",
        "Phase III→approval",
        1,
        "Phase III→approval, high_vs_low (therapeutic areas): risk ratio",
        1.04,
        0.005,
    ),
    (
        "S10.31",
        "risk_ratio",
        "TAs",
        "Phase III→approval",
        2,
        "Phase III→approval, high_vs_med (therapeutic areas): risk ratio",
        0.98,
        0.005,
    ),
    ("S10.32", "group_n", "gPS", None, 0, "pairs in group gPS 1 (gPS)", 5724, 0),
    ("S10.33", "group_n", "gPS", None, 1, "pairs in group gPS 2–9 (gPS)", 9542, 0),
    ("S10.34", "group_n", "gPS", None, 2, "pairs in group gPS ≥10 (gPS)", 3214, 0),
    ("S10.35", "n_at_start", "gPS", "Phase I→II", 0, "Phase I→II, gPS 1: pairs entering", 5724, 0),
    ("S10.36", "rate", "gPS", "Phase I→II", 0, "Phase I→II, gPS 1: success (%)", 82.3, 0.05),
    ("S10.37", "n_at_start", "gPS", "Phase I→II", 1, "Phase I→II, gPS 2–9: pairs entering", 9542, 0),
    ("S10.38", "rate", "gPS", "Phase I→II", 1, "Phase I→II, gPS 2–9: success (%)", 84.4, 0.05),
    ("S10.39", "n_at_start", "gPS", "Phase I→II", 2, "Phase I→II, gPS ≥10: pairs entering", 3214, 0),
    ("S10.40", "rate", "gPS", "Phase I→II", 2, "Phase I→II, gPS ≥10: success (%)", 83.4, 0.05),
    ("S10.41", "risk_ratio", "gPS", "Phase I→II", 0, "Phase I→II, med_vs_low (gPS): risk ratio", 1.03, 0.005),
    ("S10.42", "risk_ratio", "gPS", "Phase I→II", 1, "Phase I→II, high_vs_low (gPS): risk ratio", 1.01, 0.005),
    ("S10.43", "risk_ratio", "gPS", "Phase I→II", 2, "Phase I→II, high_vs_med (gPS): risk ratio", 0.99, 0.005),
    ("S10.44", "n_at_start", "gPS", "Phase II→III", 0, "Phase II→III, gPS 1: pairs entering", 4708, 0),
    ("S10.45", "rate", "gPS", "Phase II→III", 0, "Phase II→III, gPS 1: success (%)", 56.7, 0.05),
    ("S10.46", "n_at_start", "gPS", "Phase II→III", 1, "Phase II→III, gPS 2–9: pairs entering", 8051, 0),
    ("S10.47", "rate", "gPS", "Phase II→III", 1, "Phase II→III, gPS 2–9: success (%)", 53.8, 0.05),
    ("S10.48", "n_at_start", "gPS", "Phase II→III", 2, "Phase II→III, gPS ≥10: pairs entering", 2682, 0),
    ("S10.49", "rate", "gPS", "Phase II→III", 2, "Phase II→III, gPS ≥10: success (%)", 51.6, 0.05),
    ("S10.50", "risk_ratio", "gPS", "Phase II→III", 0, "Phase II→III, med_vs_low (gPS): risk ratio", 0.95, 0.005),
    ("S10.51", "risk_ratio", "gPS", "Phase II→III", 1, "Phase II→III, high_vs_low (gPS): risk ratio", 0.91, 0.005),
    ("S10.52", "risk_ratio", "gPS", "Phase II→III", 2, "Phase II→III, high_vs_med (gPS): risk ratio", 0.96, 0.005),
    ("S10.53", "n_at_start", "gPS", "Phase III→approval", 0, "Phase III→approval, gPS 1: pairs entering", 2671, 0),
    ("S10.54", "rate", "gPS", "Phase III→approval", 0, "Phase III→approval, gPS 1: success (%)", 29.0, 0.05),
    ("S10.55", "n_at_start", "gPS", "Phase III→approval", 1, "Phase III→approval, gPS 2–9: pairs entering", 4335, 0),
    ("S10.56", "rate", "gPS", "Phase III→approval", 1, "Phase III→approval, gPS 2–9: success (%)", 31.7, 0.05),
    ("S10.57", "n_at_start", "gPS", "Phase III→approval", 2, "Phase III→approval, gPS ≥10: pairs entering", 1383, 0),
    ("S10.58", "rate", "gPS", "Phase III→approval", 2, "Phase III→approval, gPS ≥10: success (%)", 29.4, 0.05),
    (
        "S10.59",
        "risk_ratio",
        "gPS",
        "Phase III→approval",
        0,
        "Phase III→approval, med_vs_low (gPS): risk ratio",
        1.09,
        0.005,
    ),
    (
        "S10.60",
        "risk_ratio",
        "gPS",
        "Phase III→approval",
        1,
        "Phase III→approval, high_vs_low (gPS): risk ratio",
        1.01,
        0.005,
    ),
    (
        "S10.61",
        "risk_ratio",
        "gPS",
        "Phase III→approval",
        2,
        "Phase III→approval, high_vs_med (gPS): risk ratio",
        0.93,
        0.005,
    ),
]
GROUPS = {"TAs": ["1 TA", "2–5 TAs", "≥6 TAs"], "gPS": ["gPS 1", "gPS 2–9", "gPS ≥10"]}

for identifier, kind, grouping, transition, index, _description, _expected, _tolerance in IDS:
    if kind == "universe":
        continue
    label = GROUPS[grouping][index]
    if kind == "group_n":
        row = rates[grouping][rates[grouping]["group"] == label].iloc[0]
        numbers[identifier] = int(row["group_n"])
    elif kind in ("n_at_start", "rate"):
        row = rates[grouping][(rates[grouping]["group"] == label) & (rates[grouping]["transition"] == transition)].iloc[
            0
        ]
        numbers[identifier] = int(row["n_at_start"]) if kind == "n_at_start" else round(100 * float(row["rate"]), 1)
    else:
        row = tests[grouping][
            (tests[grouping]["transition"] == transition)
            & (tests[grouping]["comparison_key"] == ["med_vs_low", "high_vs_low", "high_vs_med"][index])
        ].iloc[0]
        numbers[identifier] = round(float(row["risk_ratio"]), 2)

comparison = pd.DataFrame(
    [
        {
            "id": identifier,
            "claim": description,
            "published": expected,
            "computed": numbers[identifier],
            "matches": abs(numbers[identifier] - expected) <= tolerance,
        }
        for identifier, _kind, _grouping, _transition, _index, description, expected, tolerance in IDS
    ]
)
print(f"cells reproducing the published tables: {int(comparison['matches'].sum())} of {len(comparison)}")
comparison[~comparison["matches"]]
```
